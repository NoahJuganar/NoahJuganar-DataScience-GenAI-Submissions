# # 1. Univariate Analysis
We will individually examine the main numerical and categorical variables to understand their distributions, summary statistics, and key characteristics.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Configure visual aesthetics
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (12, 5)

# 1. Numerical Variables: Age, SibSp, Parch
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.histplot(df['Age'].dropna(), bins=20, kde=True, ax=axes[0], color='skyblue')
axes[0].set_title('Distribution of Passenger Age')
axes[0].set_xlabel('Age')

sns.countplot(data=df, x='SibSp', ax=axes[1], palette='Blues_d')
axes[1].set_title('Distribution of SibSp (Siblings/Spouses)')
axes[1].set_xlabel('SibSp count')

sns.countplot(data=df, x='Parch', ax=axes[2], palette='Greens_d')
axes[2].set_title('Distribution of Parch (Parents/Children)')
axes[2].set_xlabel('Parch count')

plt.tight_layout()
plt.show()

print("Descriptive Statistics for Numerical Features:")
display(df[['Age', 'Fare', 'SibSp', 'Parch']].describe())
```

### Interpretation of Numerical Distributions:
* **Age**: Displays a roughly normal distribution centered around 28-30 years, with a notable peak for young infants (0-5 years), indicating a diverse demographic profile.
* **SibSp & Parch**: Both display highly right-skewed distributions. The vast majority of passengers traveled alone or with only 1 family member, while large family structures on board were rare.

```python
# 2. Categorical Variables: Survived, Sex, Embarked
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.countplot(data=df, x='Survived', ax=axes[0], palette='Set2')
axes[0].set_xticklabels(['No (0)', 'Yes (1)'])
axes[0].set_title('Survival Distribution')

sns.countplot(data=df, x='Sex', ax=axes[1], palette='pastel')
axes[1].set_title('Gender Distribution')

sns.countplot(data=df, x='Embarked', ax=axes[2], palette='Set3')
axes[2].set_title('Port of Embarkation')

plt.tight_layout()
plt.show()

print("Categorical Value Counts:")
for col in ['Survived', 'Sex', 'Embarked']:
    print(f"\n--- Frequency table for {col} ---")
    display(df[col].value_counts(dropna=False))
```

### Interpretation of Categorical Distributions:

# # 2. Bivariate Analysis
In this section, we'll explore multi-variable interactions, with a strong focus on factors influencing passenger survival rates.

```python
# 1 & 2. Survival rate by Passenger Class and Sex
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Survival by Sex
sns.barplot(data=df, x='Sex', y='Survived', ax=axes[0], errorbar=None, palette='pastel')
axes[0].set_title('Survival Rate by Sex')
axes[0].set_ylabel('Survival Rate (Proportion)')

# Survival by Class and Sex
sns.barplot(data=df, x='Pclass', y='Survived', hue='Sex', ax=axes[1], errorbar=None, palette='muted')
axes[1].set_title('Survival Rate by Class and Sex')
axes[1].set_ylabel('Survival Rate (Proportion)')

plt.tight_layout()
plt.show()

# Pivot table with counts to assess sample size robustness
pivot_sex_pclass = df.pivot_table(index=['Pclass', 'Sex'], values='Survived', aggfunc=['mean', 'count'])
pivot_sex_pclass.columns = ['Survival Rate', 'Passenger Count']
pivot_sex_pclass['Survival Rate'] = (pivot_sex_pclass['Survival Rate'] * 100).round(2).astype(str) + '%'
display(pivot_sex_pclass)
```

# # 3. Outlier and Extreme-Value Analysis
We will now use boxplots and the statistical Interquartile Range (IQR) method to explore outliers in `Fare` and `Age` variables.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.boxplot(data=df, y='Fare', ax=axes[0], color='lightcoral')
axes[0].set_title('Boxplot of Fares')

sns.boxplot(data=df, y='Age', ax=axes[1], color='lightblue')
axes[1].set_title('Boxplot of Ages')

plt.show()

# Identify numerical boundaries using the 1.5 * IQR rule
for col in ['Fare', 'Age']:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    print(f"\n{col} IQR Bounds: Lower = {lower_bound:.2f}, Upper = {upper_bound:.2f}")
    print(f"Number of statistical outliers identified: {len(outliers)} ({len(outliers)/len(df)*100:.2f}% of data)")

# Investigate anomalous zero fares
zero_fares = df[df['Fare'] == 0]
print(f"\nNumber of passengers with a $0 Fare: {len(zero_fares)}")
```

### Interpretation of Outliers:
* **Fares**: Over 13% of passengers fall outside the statistical upper bound ($65.63). However, these are legitimate luxury first-class tickets (maxing out at $512.33) rather than data entry errors. Removing them would introduce major bias.
* **Ages**: Extreme values are minor, mainly pointing to elderly passengers (max 80). These are plausible ages.
* **Anomalous Zero Fares**: 15 passengers registered a $0 fare. This is likely a subset of ship employees, promotional tickets, or missing billing information.

# # 4. Missing-Value Analysis and Cleaning
We will analyze the missing values and clean them safely on a separate DataFrame copy (`df_clean`).

```python
# Create missing values summary
missing_count = df.isnull().sum()
missing_percent = (df.isnull().sum() / len(df)) * 100
missing_summary = pd.DataFrame({'Missing Count': missing_count, 'Percentage (%)': missing_percent.round(2)})
display(missing_summary[missing_summary['Missing Count'] > 0])

# Apply robust imputations on a separate copy
df_clean = df.copy()

# 1. Fill Embarked with its mode ('S')
mode_embarked = df_clean['Embarked'].mode()[0]
df_clean['Embarked'] = df_clean['Embarked'].fillna(mode_embarked)

# 2. Fill Cabin with 'Unknown'
df_clean['Cabin'] = df_clean['Cabin'].fillna('Unknown')

# 3. Fill Age with class-specific & sex-specific median values (rather than global median)
age_medians = df_clean.groupby(['Pclass', 'Sex'])['Age'].transform('median')
df_clean['Age'] = df_clean['Age'].fillna(age_medians)

# Confirm that missing values have been handled successfully
print("\nRemaining missing values in df_clean:")
print(df_clean.isnull().sum())
```

### Missing-Data Strategy & Reflection:
* **Cabin**: Since 77.1% is missing, any imputation would be arbitrary. Labeling it 'Unknown' preserves information without bias.
* **Age**: Grouping by `Pclass` and `Sex` yields significantly more accurate age estimations compared to global mean/median substitution.

# # 5. Feature Engineering
We will extract new indicators to evaluate patterns regarding group dynamics and titles.

```python
# 1. FamilySize
df_clean['FamilySize'] = df_clean['SibSp'] + df_clean['Parch'] + 1

# 2. IsAlone
df_clean['IsAlone'] = (df_clean['FamilySize'] == 1).astype(int)

# 3. Title Extraction
df_clean['Title'] = df_clean['Name'].str.extract(' ([A-Za-z]+)\.', expand=False)
# Group rare titles
df_clean['Title'] = df_clean['Title'].replace(['Lady', 'Countess','Capt', 'Col', 'Don', 'Dr', 'Major', 'Rev', 'Sir', 'Jonkheer', 'Dona'], 'Rare')
df_clean['Title'] = df_clean['Title'].replace('Mlle', 'Miss')
df_clean['Title'] = df_clean['Title'].replace('Ms', 'Miss')
df_clean['Title'] = df_clean['Title'].replace('Mme', 'Mrs')

# Visualize Engineered Feature Distributions with Survival
fig, axes = plt.subplots(1, 2, figsize=(15, 5))
sns.barplot(data=df_clean, x='FamilySize', y='Survived', ax=axes[0], palette='Blues_d', errorbar=None)
axes[0].set_title('Survival Rate by Family Size')

sns.barplot(data=df_clean, x='Title', y='Survived', ax=axes[1], palette='Set2', errorbar=None)
axes[1].set_title('Survival Rate by Title')

plt.show()
```

### Interpretation of Engineered Features:
* **FamilySize**: Mid-sized families (2 to 4 people) had substantially higher survival rates. Larger families (5+) or those traveling completely alone had poor survival chances.
* **Titles**: "Mrs" and "Miss" titles exhibit the highest survival rates, validating social patterns of prioritization.

# # 6. Final Report & Conclusions

### Key Findings
1. **Socio-Economic Stratification**: Survival rates are heavily stratified by ticket class. Class 1 has a **63.0%** survival rate compared to only **24.2%** in Class 3.
2. **Gender Bias**: Females survived at vastly higher rates (**74.2%** overall) compared to males (**18.9%**).
3. **Demographic Influence**: Infants and young children (ages 0-5) were prioritized and had elevated rates of survival.
4. **Family Dynamics**: Traveling alone (`IsAlone`) or with huge families (size 5+) significantly reduced survival probability compared to having 1-3 family companions.
5. **Port Influence**: Cherbourg (C) embarkation experienced higher average fares and higher relative survival rates, while Southampton (S) passengers suffered lower relative survival rates.

### Data Quality & Limitations
* **Missing Values**: We identified significant missing value issues, notably in the `Cabin` column (77% missing) and `Age` (20% missing). While group-based median imputation can alleviate missingness in `Age`, conclusions based on imputed values must be treated with caution.
* **Sample Constraints**: This dataset of 891 records represents only a subset of the actual passengers on board the Titanic.

### Final Conclusion
Our exploratory analysis conclusively shows that **class and sex** were the strongest structural determinants of survival during the Titanic disaster. The data supports the historical account of the evacuation order prioritizing "women and children", but reveals that this protocol was heavily influenced by the class status of the passenger.
