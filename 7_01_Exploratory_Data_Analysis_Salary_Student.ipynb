### Visualising the Data

**Task 8 — Bar chart.** Create a bar chart comparing the **average salary in USD** across experience levels (`EN`, `MI`, `SE`, `EX`). Add clear axis labels and a title. Which group has the highest average?

```python

# Calculate average salary by experience level
avg_salary = df.groupby("experience_level")["salary_in_usd"].mean()

# Arrange experience levels in logical order
avg_salary = avg_salary.reindex(["EN", "MI", "SE", "EX"])

# Create bar chart
plt.figure(figsize=(8, 5))
plt.bar(avg_salary.index, avg_salary.values)

plt.xlabel("Experience Level")
plt.ylabel("Average Salary (USD)")
plt.title("Average Salary by Experience Level")
plt.show()

# Display exact averages
print(avg_salary.round(2))
```

**Task 9 — Scatter plot.** Make a scatter plot of `work_year` against `salary_in_usd`. Label the axes. What patterns do you notice? Are any points difficult to distinguish because they overlap?

```python

plt.figure(figsize=(9, 5))

plt.scatter(
    df["work_year"],
    df["salary_in_usd"],
    alpha=0.15,
    s=10
)

plt.xlabel("Work Year")
plt.ylabel("Salary (USD)")
plt.title("Salary Distribution by Work Year")
plt.xticks(sorted(df["work_year"].unique()))

plt.show()
```

**Task 10 — Distribution.** Plot the distribution of `salary_in_usd`. Does it look approximately normal, skewed, or something else? Are there extreme values?

```python

plt.figure(figsize=(9, 5))

sns.histplot(
    df["salary_in_usd"],
    bins=40,
    kde=True
)

plt.xlabel("Salary (USD)")
plt.ylabel("Number of Observations")
plt.title("Distribution of Salaries in USD")

plt.show()
```

**Task 11 — Pairwise relationships.** Visualize pairwise relationships among `salary_in_usd`, `work_year`, and `remote_ratio`, including distributions of the individual variables.

Try to compare the variables in one combined visualization. What relationships or patterns do you notice?

```python

sns.pairplot(
    df[["salary_in_usd", "work_year", "remote_ratio"]],
    diag_kind="hist",
    corner=True,
    plot_kws={"alpha": 0.15, "s": 10}
)

plt.show()
```

**Task 12 — Remove a column.** Remove the original-currency `salary` column while retaining `salary_in_usd`, then inspect the updated DataFrame. Why might you choose to keep one salary column rather than both?

### Reflection

The dataset contains 16,534 salary observations across 11 variables. Key findings:

- **Data quality:** No missing values, but 6,421 duplicate records (38.8%). These may represent different employees with identical characteristics, so they should not automatically be removed.
- **Salary distribution:** Salaries are right-skewed, with a mean of $149,687 exceeding the median of $141,300. Some high salaries reach $800,000.
- **Experience:** Average salaries increase with experience, from approximately $92,327 for entry-level employees to $195,264 for executives.
- **Other patterns:** Most positions are either fully remote or on-site. The dataset is heavily concentrated in 2023–2024, making trends over time difficult to establish.

**Before machine learning**, I would investigate potential outliers, examine salary differences by job title and location, assess whether duplicates are legitimate, and check for imbalances between categories. Categorical variables would need encoding, and appropriate train/test splitting would be necessary to avoid data leakage.

Overall, EDA helps us understand data quality, identify patterns and determine what preparation is required before modelling.
